````python
# ==============================================================================
# Exemplos de Manipulação de Arquivos em Python (Nível Iniciante)
# ==============================================================================

# Exemplo 1: Escrever em um arquivo de texto e depois ler seu conteúdo

# Define o nome do arquivo
nome_arquivo = "teste de aula.txt" #Este arquivo existe, por enquanto, somente dentro do ambiente de execução

# Modo 'w' (write): Abre o arquivo para escrita. Se o arquivo existir,
#seu conteúdo será apagado. Se não existir, será criado.
# O 'with open(...)' garante que o arquivo será fechado automaticamente,
# mesmo que ocorra um erro.
with open(nome_arquivo, "w", encoding="utf-8") as arquivo:
    # Escreve linhas no arquivo. O caractere '\n' representa uma quebra de linha.
    arquivo.write("Nome: Fulano de tal\n")
    arquivo.write("Idade: 21 anos\n")
    arquivo.write("Profissão: Estudante\n")

print(f"Conteúdo escrito em '{nome_arquivo}'")
````

````python
# Versão do código acima que solicita o nome do arquivo e
# conteúdo ao usuário, via função input
nome_arquivo2 = input("Digite o nome do arquivo") +".txt"

with open(nome_arquivo2, "w") as arquivo2:
    arquivo2.write(input("Digite o conteúdo do arquivo: "))
print(f"Conteúdo escrito em '{nome_arquivo2}'")
````

````python
# Modo 'r' (read): Abre o arquivo para leitura.
with open(nome_arquivo2, "r", encoding="utf-8") as arquivo:
    # Lê todo o conteúdo do arquivo.
    conteudo = arquivo.read()
    print(f"\nConteúdo lido de '{nome_arquivo2}':")
    print()
    print(conteudo)
````

````python
# Exemplo 2: Adicionar conteúdo a um arquivo existente

# Modo 'a' (append): Abre o arquivo para escrita.
#Se o arquivo existir, o novo conteúdo será adicionado ao final.
#Se não existir, será criado.
with open(nome_arquivo2, "a", encoding="utf-8") as arquivo:
    # Adiciona mais uma linha ao final do arquivo.
    arquivo.write("Nome: José Maria\n")
    arquivo.write("Idade: 23 anos\n")
    arquivo.write("Profissão: Diretor\n")

print(f"Conteúdo adicionado a '{nome_arquivo2}'")
````

````python
# Lê novamente o conteúdo para ver a linha adicionada
with open(nome_arquivo2, "r", encoding="utf-8") as arquivo:
    conteudo_atualizado = arquivo.read()
    print(f"\nConteúdo atualizado de '{nome_arquivo2}':")
    print(conteudo_atualizado)
````

````python
# Para baixar para a máquina ou servidor local o arquivo criado dentro do ambiente de execução,
# permitindo ao usuário selecionar o arquivo.

from google.colab import files
import os

# Obtém a lista de arquivos no diretório atual (normalmente /content/ no Colab)
# Filtra para incluir apenas arquivos (excluindo diretórios)
arquivos_disponiveis = [f for f in os.listdir('.') if os.path.isfile(f)]

if not arquivos_disponiveis:
    print("Não há arquivos disponíveis para download no diretório atual.")
else:
    print("Arquivos disponíveis para download:")
    # Mostra a lista de arquivos numerada para o usuário
    for i, arquivo in enumerate(arquivos_disponiveis):
        print(f"{i + 1}: {arquivo}")

    try:
        # Solicita ao usuário que escolha um arquivo
        escolha = input(f"Digite o número do arquivo que deseja baixar (1 a {len(arquivos_disponiveis)}): ")
        indice_escolhido = int(escolha) - 1 # Converte a escolha para índice baseado em 0

        # Verifica se o índice escolhido é válido
        if 0 <= indice_escolhido < len(arquivos_disponiveis):
            nome_do_arquivo_para_baixar = arquivos_disponiveis[indice_escolhido]

            try:
                files.download(nome_do_arquivo_para_baixar)
                print(f"Arquivo '{nome_do_arquivo_para_baixar}' solicitado para download.")
            except FileNotFoundError:
                print(f"Erro: Arquivo '{nome_do_arquivo_para_baixar}' não encontrado no ambiente Colab (apesar de estar na lista).")
            except Exception as e:
                print(f"Ocorreu um erro durante o download: {e}")
        else:
            print("Escolha inválida. Por favor, digite um número da lista.")

    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")
    except Exception as e:
        print(f"Ocorreu um erro inesperado durante o processo de seleção: {e}")
````

O trecho de código: arquivos_disponiveis = [f for f in os.listdir('.') if os.path.isfile(f)] é um exemplo de uma list comprehension (compreensão de lista) em Python.

Ele faz o seguinte:

os.listdir('.'): Esta parte lista todos os itens (arquivos e diretórios) presentes no diretório atual (representado por '.').
for f in os.listdir('.'): Isso itera sobre cada item retornado por os.listdir('.'). Para cada item, ele o atribui temporariamente à variável f.
if os.path.isfile(f): Esta é uma condição. Para cada item f, a função os.path.isfile(f) verifica se f é um arquivo (e não um diretório ou outro tipo de item do sistema de arquivos). A condição é verdadeira se for um arquivo.
f: Se a condição if os.path.isfile(f) for verdadeira, o item f (que é o nome do arquivo) é incluído na nova lista que está sendo criada.
[...]: Os colchetes indicam que o resultado de toda a operação é uma nova lista.
Portanto, a linha inteira constrói uma nova lista chamada arquivos_disponiveis que contém apenas os nomes dos arquivos (excluindo diretórios) presentes no diretório atual. Isso é útil para garantir que a lista mostrada ao usuário e usada para download contenha apenas itens que podem ser baixados como arquivos.

````python
# Permite ao usuário escolher o modo de abertura do arquivo ('w', 'r' ou 'a')
# Encapsulado em uma função

import os
from google.colab import files # Import files para download

def manipular_arquivo():

    nome_arquivo =input("Digite o nome do arquivo") + ".txt" # Usando o nome de arquivo já definido anteriormente

    # Solicita ao usuário o modo desejado ou ação
    modo_escolhido = input(f"Escolha o modo para abrir '{nome_arquivo}' ('w' para escrever, 'r' para ler, 'a' para adicionar, 'd' para download, 'x' para excluir): ").lower()

    # --- Camada Simples de Cibersegurança ---
    # Validar a entrada do usuário para garantir que é um modo permitido.
    modos_permitidos = ['w', 'r', 'a', 'd', 'x']

    if modo_escolhido not in modos_permitidos:
        print(f"Modo '{modo_escolhido}' inválido. Escolha entre {modos_permitidos}.")
    else:
        try:
            if modo_escolhido in ['w', 'r', 'a']:
                # Abre o arquivo nos modos de manipulação de conteúdo
                with open(nome_arquivo, modo_escolhido, encoding="utf-8") as arquivo:
                    print(f"\nArquivo '{nome_arquivo}' aberto no modo '{modo_escolhido}'.")

                    if modo_escolhido == 'r':
                        # Modo leitura: lê e exibe o conteúdo
                        conteudo = arquivo.read()
                        print(f"Conteúdo lido:\n{conteudo}")
                    elif modo_escolhido == 'w':
                        # Modo escrita: solicita e escreve novo conteúdo
                        novo_conteudo = input("Digite o novo conteúdo para o arquivo (sobrescreverá o existente): ")
                        arquivo.write(novo_conteudo + "\n")
                        print("Conteúdo escrito no arquivo.")
                    elif modo_escolhido == 'a':
                        # Modo adicionar: solicita conteúdo e adiciona ao final
                        conteudo_para_adicionar = input("Digite o conteúdo para adicionar ao arquivo: ")
                        arquivo.write(conteudo_para_adicionar + "\n")
                        print("Conteúdo adicionado ao arquivo.")

            elif modo_escolhido == 'd':
                # Modo download
                if os.path.exists(nome_arquivo):
                    try:
                        files.download(nome_arquivo)
                        print(f"Arquivo '{nome_arquivo}' solicitado para download.")
                    except Exception as e:
                        print(f"Ocorreu um erro durante o download: {e}")
                else:
                     print(f"Erro: Arquivo '{nome_arquivo}' não encontrado para download.")

            elif modo_escolhido == 'x':
                # Modo excluir
                if os.path.exists(nome_arquivo):
                    try:
                        os.remove(nome_arquivo)
                        print(f"Arquivo '{nome_arquivo}' excluído com sucesso.")
                    except Exception as e:
                        print(f"Ocorreu um erro ao tentar excluir o arquivo: {e}")
                else:
                    print(f"Erro: Arquivo '{nome_arquivo}' não encontrado para exclusão.")


        except FileNotFoundError:
            # Este erro geralmente só ocorre no modo 'r' se o arquivo não existir
            print(f"Erro: Arquivo '{nome_arquivo}' não encontrado no modo de leitura ('r').")
        except IOError as e:
            print(f"Erro de E/S ao manipular o arquivo: {e}")
        except Exception as e:
            print(f"Ocorreu um erro inesperado: {e}")
````

````python
#guardando o arquivo diretamente em diretório do Google Drive

import os

# Define o nome do arquivo que você quer salvar
nome_arquivo_para_salvar = "teste de aula.txt" # Ou o nome do arquivo que você criou

# Define o caminho completo para salvar no Google Drive
# Substitua 'Meu Drive' pelo nome da sua pasta raiz no Google Drive, se for diferente.
# Você pode criar subpastas se desejar, por exemplo: '/content/drive/MyDrive/MinhaPasta/arquivo.txt'
caminho_google_drive = f"/content/drive/MyDrive/{nome_arquivo_para_salvar}"

# --- Camada Simples de Cibersegurança ---
# Verificar se a pasta de destino no Drive existe antes de tentar salvar
diretorio_destino = os.path.dirname(caminho_google_drive)
if not os.path.exists(diretorio_destino):
    print(f"Diretório '{diretorio_destino}' não encontrado no Google Drive. Criando...")
    try:
        os.makedirs(diretorio_destino)
        print("Diretório criado com sucesso.")
    except Exception as e:
        print(f"Erro ao criar diretório no Google Drive: {e}")
        # Se não conseguir criar o diretório, talvez seja melhor parar ou tentar um caminho diferente.
        # Por simplicidade, este exemplo continua, mas em um script real você poderia adicionar mais tratamento de erro.


# Verifica se o arquivo original existe no ambiente Colab antes de tentar copiá-lo
if os.path.exists(nome_arquivo_para_salvar):
    print(f"\nSalvando '{nome_arquivo_para_salvar}' para o Google Drive em '{caminho_google_drive}'")
    try:
        # Abre o arquivo original para leitura e o arquivo de destino no Drive para escrita
        with open(nome_arquivo_para_salvar, 'r', encoding='utf-8') as f_origem, \
             open(caminho_google_drive, 'w', encoding='utf-8') as f_destino:
            # Lê o conteúdo do arquivo original
            conteudo = f_origem.read()
            # Escreve o conteúdo no arquivo no Google Drive
            f_destino.write(conteudo)

        print("Arquivo salvo com sucesso no Google Drive.")

    except IOError as e:
        print(f"Erro de E/S ao salvar o arquivo no Google Drive: {e}")
    except Exception as e:
        print(f"Ocorreu um erro inesperado ao salvar no Drive: {e}")
else:
    print(f"Erro: Arquivo '{nome_arquivo_para_salvar}' não encontrado no ambiente Colab para ser salvo.")
````

'''O trecho de código  "except Exception as e:" faz parte de um bloco try...except em Python.
Ele é usado para capturar e lidar com quaisquer erros ou exceções inesperadas que possam ocorrer durante a execução do código dentro do bloco try.

except Exception as e: especifica que, se qualquer tipo de exceção (erro) ocorrer no bloco try, ela será capturada. A exceção capturada é atribuída à variável e
pode ser usada para acessar informações sobre o erro (como a mensagem de erro).
print(f"Ocorreu um erro durante o download: {e}"): Esta linha dentro do bloco except será executada se uma exceção for capturada.
Ela imprime uma mensagem informativa indicando que ocorreu um erro durante o download e inclui a mensagem de erro específica contida na variável e.
Em resumo, este bloco garante que, se algo der errado durante a tentativa de download do arquivo (além do arquivo não ser encontrado, que é tratado pelo except FileNotFoundError),
o programa não trave, mas sim capture o erro, imprima uma mensagem e continue a execução (ou encerre de forma controlada).'''

````python
# --- Camada Simples de Cibersegurança para Manipulação de Arquivos ---

#Exemplo 1:
# Verificar a existência do arquivo antes de tentar manipulá-lo para evitar erros.
# Usar 'with open()' para garantir que o arquivo seja sempre fechado, liberando recursos.
# Especificar a codificação ('utf-8') para evitar problemas com caracteres especiais.

import os

nome_arquivo_seguro = "arquivo_seguro.txt"

# Exemplo de verificação de existência antes de tentar ler
if os.path.exists(nome_arquivo_seguro):
    print(f"\nVerificação de segurança: Arquivo '{nome_arquivo_seguro}' existe.")
    try:
        with open(nome_arquivo_seguro, "r", encoding="utf-8") as arquivo:
            conteudo_seguro = arquivo.read()
            print(f"Conteúdo lido de '{nome_arquivo_seguro}':\n{conteudo_seguro}")
    except IOError as e:
        print(f"Erro de E/S ao ler o arquivo: {e}")
else:
    print(f"\nVerificação de segurança: Arquivo '{nome_arquivo_seguro}' não encontrado. Criando...")
    try:
        with open(nome_arquivo_seguro, "w", encoding="utf-8") as arquivo:
            arquivo.write("Conteúdo inicial do arquivo seguro.\n")
            print(f"Arquivo '{nome_arquivo_seguro}' criado com sucesso.")
    except IOError as e:
        print(f"Erro de E/S ao criar o arquivo: {e}")
````

````python
#Exemplo 2: Este código "simulará" a necessidade da senha para ler e escrever no arquivo.

#Passos para usar este exemplo:
#Abra o Gerenciador de Secrets no painel esquerdo (ícone de chave).
#Clique em "Adicionar Secret".
#No campo "Name", digite FILE_PASSWORD.
#No campo "Value", digite a senha que você deseja usar.
#Execute a célula de código abaixo.


from google.colab import userdata
import os

# Define o nome do arquivo
nome_arquivo_secreto = "arquivo_com_secret.txt"

# Obtém a senha do Gerenciador de Secrets do Colab
# Certifique-se de ter adicionado um Secret com o nome 'FILE_PASSWORD'
try:
    senha_secreta = userdata.get('FILE_PASSWORD')
    if senha_secreta is None:
        print("Erro: Secret 'FILE_PASSWORD' não encontrado. Adicione-o no Gerenciador de Secrets.")
except Exception as e:
    print(f"Erro ao acessar o Secret: {e}")
    senha_secreta = None # Define como None se houver erro ao obter o secret


# Função para simular login e manipular o arquivo
def manipular_arquivo_com_senha(senha_fornecida):
    if senha_secreta is None:
        print("\nOperação cancelada: Senha secreta não disponível.")
        return

    # Verifica se a senha fornecida corresponde à senha armazenada no Secret
    if senha_fornecida == senha_secreta:
        print("\nSenha correta. Acessando o arquivo...")
        try:
            # Verifica se o arquivo existe
            if os.path.exists(nome_arquivo_secreto):
                print(f"Arquivo '{nome_arquivo_secreto}' encontrado.")
                # Lê o conteúdo atual
                with open(nome_arquivo_secreto, "r", encoding="utf-8") as arquivo:
                    conteudo = arquivo.read()
                    print(f"Conteúdo atual:\n{conteudo}")

                # Adiciona uma nova linha (simulando escrita)
                with open(nome_arquivo_secreto, "a", encoding="utf-8") as arquivo:
                    nova_linha = f"Esta linha foi adicionada com a senha correta em {os.path.getmtime(nome_arquivo_secreto)}.\n"
                    arquivo.write(nova_linha)
                    print("Nova linha adicionada.")

                # Lê o conteúdo atualizado
                with open(nome_arquivo_secreto, "r", encoding="utf-8") as arquivo:
                    conteudo_atualizado = arquivo.read()
                    print(f"Conteúdo atualizado:\n{conteudo_atualizado}")

            else:
                # Cria o arquivo se não existir
                print(f"Arquivo '{nome_arquivo_secreto}' não encontrado. Criando...")
                with open(nome_arquivo_secreto, "w", encoding="utf-8") as arquivo:
                    arquivo.write("Conteúdo inicial protegido por senha.\n")
                    print(f"Arquivo '{nome_arquivo_secreto}' criado com sucesso.")

        except IOError as e:
            print(f"Erro de E/S ao manipular o arquivo: {e}")
        except Exception as e:
             print(f"Ocorreu um erro inesperado: {e}")

    else:
        print("\nSenha incorreta. Acesso negado.")

````

Aqui estão alguns exemplos de ferramentas mais adequadas, com um foco didático:

Variáveis de Ambiente (Environment Variables):
O que são: São variáveis definidas fora do seu código, geralmente no nível do sistema operacional ou do ambiente de execução (como no terminal ou em um servidor).
Seu código acessa esses valores sem tê-los escritos diretamente.
Por que são úteis didaticamente: Introduzem o conceito fundamental de separar configuração (incluindo segredos) do código-fonte. É uma prática muito comum em desenvolvimento web e de aplicações.

````python

Como usar (exemplo básico em Python):
import os

# Para definir uma variável de ambiente no terminal (Linux/macOS):
# export MINHA_SENHA="senha_super_secreta"

# No código Python, você a acessa assim:
senha = os.environ.get('MINHA_SENHA')

if senha:
    print("Senha acessada via variável de ambiente.")
    # Use a senha...
else:
    print("Variável de ambiente 'MINHA_SENHA' não encontrada.")
Limitações: Não são criptografadas por padrão e podem ser acessíveis a outros processos no mesmo sistema se não forem configuradas corretamente.
Arquivos de Configuração (com Criptografia Simples):
O que são: Armazenar segredos em arquivos separados do código principal. Para segurança, esses arquivos podem ser criptografados.
Por que são úteis didaticamente: Demonstram a importância de não "hardcodar" (escrever diretamente no código) informações sensíveis e introduzem a ideia de criptografia para proteção.
Como usar (conceito - requer bibliotecas de criptografia): Você pode usar bibliotecas Python como cryptography para criptografar e descriptografar um arquivo contendo suas senhas.
# Exemplo conceitual - requer a biblioteca 'cryptography'
# pip install cryptography

from cryptography.fernet import Fernet

# Gerar uma chave (faça isso apenas uma vez e guarde a chave em local seguro,
# talvez outra variável de ambiente ou gerenciador de segredos!)
# key = Fernet.generate_key()
# print(key) # Guarde esta chave!

# Exemplo de uso com uma chave existente
key = b'SUA_CHAVE_GERADA_AQUI==' # Substitua pela chave real
cipher_suite = Fernet(key)

# Conteúdo a ser criptografado
conteudo_secreto = b"minha_senha_do_banco=12345\nminha_chave_api=XYZABC"

# Criptografar
conteudo_criptografado = cipher_suite.encrypt(conteudo_secreto)

# Salvar em arquivo (arquivo_secreto.encrypted)
with open("arquivo_secreto.encrypted", "wb") as f:
    f.write(conteudo_criptografado)

print("Conteúdo criptografado e salvo.")

# Para ler e descriptografar
with open("arquivo_secreto.encrypted", "rb") as f:
    conteudo_lido_criptografado = f.read()

try:
    conteudo_descriptografado = cipher_suite.decrypt(conteudo_lido_criptografado).decode('utf-8')
    print("\nConteúdo lido e descriptografado:")
    print(conteudo_descriptografado)
    # Agora você pode parsear este conteúdo (ex: splitlines(), split('='))
except Exception as e:
    print(f"Erro ao descriptografar: {e}")

Limitações: A chave de criptografia em si se torna um segredo que precisa ser gerenciado de forma segura.
Gerenciadores de Secrets Dedicados (Ex: HashiCorp Vault, AWS Secrets Manager, Azure Key Vault):
O que são: Serviços e softwares profissionais projetados especificamente para armazenar, gerenciar e distribuir segredos de forma segura (senhas, chaves de API, certificados, etc.).
Por que são úteis didaticamente: Embora mais complexos para iniciantes, mencioná-los mostra o "próximo nível" em segurança de segredos em ambientes de produção e a importância de soluções centralizadas.
Como usar: Envolve configurar o serviço, definir políticas de acesso e usar APIs ou SDKs para acessar os segredos em suas aplicações.
Limitações: Geralmente requerem configuração de infraestrutura e são mais complexos para configurar e usar em comparação com variáveis de ambiente ou arquivos simples.
Qual escolher para fins didáticos?

Variáveis de Ambiente são um excelente ponto de partida. São fáceis de entender e configurar para exemplos básicos e demonstram um princípio de segurança fundamental.
Arquivos de Configuração com Criptografia Simples são bons para demonstrar a ideia de proteger os dados em repouso, mas a gestão da chave de criptografia adiciona uma camada de complexidade.
Para os exemplos que você já tem, refatorar a parte que usa o userdata.get() para usar variáveis de ambiente seria uma progressão natural e didática, mostrando como o código pode acessar o segredo de uma fonte externa diferente do Secrets Manager do Colab.

Gostaria que eu mostrasse como reescrever um dos seus exemplos para usar variáveis de ambiente em vez do Secrets Manager?
````

````python
#Exemplo 2: Este código "simulará" a necessidade da senha para ler e escrever no arquivo,
#agora utilizando variáveis de ambiente.

#Para usar este exemplo:
#Execute a célula acima para definir a variável de ambiente FILE_PASSWORD.

import os

# Define o nome do arquivo
nome_arquivo_secreto = "arquivo_com_secret.txt"

# Obtém a senha da variável de ambiente
try:
    senha_secreta = os.environ.get('FILE_PASSWORD')
    if senha_secreta is None:
        print("Erro: Variável de ambiente 'FILE_PASSWORD' não encontrada. Certifique-se de tê-la definido.")
except Exception as e:
    print(f"Erro ao acessar a variável de ambiente: {e}")
    senha_secreta = None # Define como None se houver erro ao obter o secret


# Função para simular login e manipular o arquivo
def manipular_arquivo_com_senha(senha_fornecida):
    if senha_secreta is None:
        print("\nOperação cancelada: Senha secreta não disponível (variável de ambiente não definida).")
        return

    # Verifica se a senha fornecida corresponde à senha armazenada na variável de ambiente
    if senha_fornecida == senha_secreta:
        print("\nSenha correta. Acessando o arquivo...")
        try:
            # Verifica se o arquivo existe
            if os.path.exists(nome_arquivo_secreto):
                print(f"Arquivo '{nome_arquivo_secreto}' encontrado.")
                # Lê o conteúdo atual
                with open(nome_arquivo_secreto, "r", encoding="utf-8") as arquivo:
                    conteudo = arquivo.read()
                    print(f"Conteúdo atual:\n{conteudo}")

                # Adiciona uma nova linha (simulando escrita)
                with open(nome_arquivo_secreto, "a", encoding="utf-8") as arquivo:
                    # Usando time.time() para uma marca de tempo mais simples, já que os.path.getmtime
                    # pode não ser atualizado imediatamente dependendo do sistema de arquivos e cache.
                    import time
                    nova_linha = f"Esta linha foi adicionada com a senha correta em {time.ctime()}.\n"
                    arquivo.write(nova_linha)
                    print("Nova linha adicionada.")

                # Lê o conteúdo atualizado
                with open(nome_arquivo_secreto, "r", encoding="utf-8") as arquivo:
                    conteudo_atualizado = arquivo.read()
                    print(f"Conteúdo atualizado:\n{conteudo_atualizado}")

            else:
                # Cria o arquivo se não existir
                print(f"Arquivo '{nome_arquivo_secreto}' não encontrado. Criando...")
                with open(nome_arquivo_secreto, "w", encoding="utf-8") as arquivo:
                    arquivo.write("Conteúdo inicial protegido por senha.\n")
                    print(f"Arquivo '{nome_arquivo_secreto}' criado com sucesso.")

        except IOError as e:
            print(f"Erro de E/S ao manipular o arquivo: {e}")
        except Exception as e:
             print(f"Ocorreu um erro inesperado: {e}")

    else:
        print("\nSenha incorreta. Acesso negado.")
````

# O que são Sockets?
# Sockets são um ponto final de comunicação em uma rede. Eles permitem que
# programas em diferentes máquinas (ou na mesma máquina) enviem e recebam dados
# uns dos outros através de protocolos de rede, como TCP/IP. Pense neles como
# as "tomadas" que permitem que diferentes computadores se "conectem" e "conversem".
#
# Para que servem?
# Sockets são a base de muitas comunicações na internet e em redes locais.
# Eles são usados para construir:
# - Servidores Web (que enviam páginas HTML)
# - Clientes de e-mail (que enviam e recebem e-mails)
# - Aplicações de chat
# - Jogos online
# - Transferência de arquivos (FTP)
# - E muitas outras aplicações que precisam trocar dados pela rede.
#
# Basicamente, sempre que dois programas precisam se comunicar por uma rede,
# eles geralmente usam sockets para estabelecer a conexão e trocar informações.

````python

#Exemplo Prático 1: Servidor local
# Nota: Para que o exemplo do socket funcione, você precisaria de um cliente e um servidor
# rodando simultaneamente. Este snippet mostra um exemplo básico de como um servidor
# simples poderia ser configurado para ouvir uma conexão e um cliente para se conectar.
# ==============================================================================

import socket
import threading
import time

# Configuração básica de um servidor socket (simplificado)

# Define o host e a porta que o servidor vai ouvir
HOST = '0.0.0.0'  # Endereço IP local (localhost)
PORT = 135        # Porta para a conexão (escolha uma porta livre)

# Função para lidar com a conexão do cliente (executada em uma thread separada)
def handle_client(conn, addr):
    print(f"Conectado por {addr}")
    # --- Camada Simples de Cibersegurança para Servidor Socket ---
    # Limitar o tempo que um cliente pode ficar conectado.
    # Limitar o tamanho dos dados recebidos para prevenir ataques de buffer overflow.
    conn.settimeout(60) # Define um timeout de 60 segundos para a conexão

    try:
        while True:
            # Recebe dados do cliente (buffer de 1024 bytes)
            # Limitar o tamanho do buffer é uma forma de segurança básica.
            data = conn.recv(1024)
            if not data:
                # Se não há dados, a conexão foi encerrada pelo cliente
                break
            # Decodifica os dados recebidos (de bytes para string)
            mensagem_recebida = data.decode('utf-8')
            print(f"Recebido de {addr}: {mensagem_recebida}")
            # Envia uma resposta de volta para o cliente (codificada para bytes)
            resposta = f"Recebi sua mensagem: {mensagem_recebida}"
            conn.sendall(resposta.encode('utf-8'))
    except socket.timeout:
        print(f"Conexão com {addr} atingiu o tempo limite.")
    except Exception as e:
        print(f"Erro ao lidar com cliente {addr}: {e}")
    finally:
        # Fecha a conexão com o cliente
        conn.close()
        print(f"Conexão com {addr} encerrada.")

# Código para iniciar o servidor (executado em uma thread separada para não bloquear o notebook)
def start_server():
    print(f"Servidor socket iniciando em {HOST}:{PORT}")
    # AF_INET: Família de endereços IPv4
    # SOCK_STREAM: Tipo de socket para TCP (Transmission Control Protocol)
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        # --- Camada Simples de Cibersegurança para Servidor Socket ---
        # Configurar a opção SO_REUSEADDR permite que o socket seja reutilizado logo após fechar.
        # Útil para evitar o erro "Address already in use" durante o desenvolvimento/testes,
        # mas deve ser usado com cautela em produção.
        s.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)

        # Liga o socket ao host e porta definidos
        s.bind((HOST, PORT))
        # Começa a ouvir por conexões de entrada (máximo de 5 conexões na fila)
        s.listen(5) # Definir um número razoável de conexões na fila
        print("Servidor ouvindo...")
        try:
            while True:
                # Aceita uma nova conexão. Retorna um novo objeto socket (conn)
                # que representa a conexão e o endereço do cliente (addr).
                conn, addr = s.accept()
                # Cria uma nova thread para lidar com o cliente, permitindo que o servidor
                # continue a aceitar outras conexões.
                thread = threading.Thread(target=handle_client, args=(conn, addr))
                thread.start()
        except Exception as e:
            print(f"Erro no servidor: {e}")
        finally:
             # Para fins de demonstração no notebook, o servidor não será executado
             # indefinidamente. Em uma aplicação real, este loop seria contínuo.
             print("Servidor encerrado.")

# Para executar o servidor, descomente a linha abaixo. Lembre-se que ele
# continuará rodando em uma thread até que você interrompa a execução da célula.
server_thread = threading.Thread(target=start_server)
server_thread.start()
# print("Thread do servidor iniciada. Para parar, interrompa a execução desta célula.")


#Configuração básica de um cliente socket (simplificado)

# Define o host e a porta do servidor ao qual o cliente vai se conectar
SERVER_HOST = '0.0.0.0'  # Endereço IP do servidor
SERVER_PORT = 135       # Porta do servidor

# Código para iniciar o cliente (executado após um pequeno atraso para permitir que o servidor inicie, se descomentado)
def start_client():
    print(f"\nCliente socket tentando conectar a {SERVER_HOST}:{SERVER_PORT}")
    time.sleep(1) # Pequeno atraso para permitir que o servidor inicie (se estiver rodando)
    try:
        # Cria um objeto socket para o cliente
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
            # --- Camada Simples de Cibersegurança para Cliente Socket ---
            # Definir um timeout para a conexão evita que o cliente fique esperando indefinidamente
            # se o servidor não responder.
            s.settimeout(10) # Timeout de 10 segundos para a conexão

            # Conecta o cliente ao servidor no host e porta especificados
            s.connect((SERVER_HOST, SERVER_PORT))
            print(f"Cliente conectado a {SERVER_HOST}:{SERVER_PORT}")
            # Mensagem a ser enviada para o servidor (codificada para bytes)
            mensagem_cliente = "Olá do cliente!"
            s.sendall(mensagem_cliente.encode('utf-8'))

            # Recebe a resposta do servidor (buffer de 1024 bytes)
            data = s.recv(1024)
            # Decodifica os dados recebidos
            resposta_servidor = data.decode('utf-8')
            print(f"Recebido do servidor: {resposta_servidor}")
    except socket.timeout:
        print("Erro no cliente: Conexão atingiu o tempo limite.")
    except ConnectionRefusedError:
        print(f"Erro: Conexão recusada. Verifique se o servidor está rodando em {SERVER_HOST}:{SERVER_PORT}.")
    except Exception as e:
        print(f"Erro no cliente: {e}")
    finally:
        print("Cliente encerrado.")

# Para executar o cliente, descomente a linha abaixo. Lembre-se que ele
# só funcionará se o servidor estiver rodando e acessível.
start_client()
````

````python
# ==============================================================================
# Exemplos prático 2: Requisições Web (HTTP) em Python (Nível Iniciante)
# ==============================================================================

import requests

# Exemplo 1: Fazer uma requisição GET simples para obter o conteúdo de uma página web

# Define a URL da página web que queremos acessar
url_exemplo = "https://autoesporte.globo.com/"

print(f"\nFazendo requisição GET para: {url_exemplo}")

try:
    # Faz a requisição GET para a URL. O método requests.get() baixa o conteúdo da URL.
    # O objeto 'response' contém informações sobre a resposta do servidor.
    # --- Camada Simples de Cibersegurança para Requisições Web ---
    # Verificar o certificado SSL/TLS é crucial para garantir que você está se conectando
    # ao servidor correto e que a comunicação é criptografada. requests faz isso por padrão.
    # Definir um timeout para a requisição evita que ela fique pendurada indefinidamente.
    response = requests.get(url_exemplo, timeout=10) # Timeout de 10 segundos

    # Verifica se a requisição foi bem-sucedida (código de status 200 indica sucesso)
    if response.status_code == 200:
        print("Requisição bem-sucedida (Status Code 200)")
        # response.text contém o conteúdo da resposta como uma string (HTML da página)
        # Exibe apenas os primeiros 500 caracteres para não sobrecarregar a saída
        print("Conteúdo da página (primeiros 500 caracteres):")
        print(response.text[:500] + "...")
    else:
        # Se o status code não for 200, algo deu errado (ex: 404 Not Found, 500 Internal Server Error)
        print(f"Erro na requisição. Status Code: {response.status_code}")

except requests.exceptions.Timeout:
    print(f"Erro na requisição: Tempo limite excedido para {url_exemplo}")
except requests.exceptions.RequestException as e:
    # Captura outras exceções que podem ocorrer durante a requisição (ex: erro de conexão, erro SSL)
    print(f"Ocorreu um erro durante a requisição: {e}")

# Exemplo 2: Fazer uma requisição GET para uma API pública que retorna dados em formato JSON

# Define a URL de uma API pública (neste caso, uma API de exemplo que retorna dados aleatórios)
# O JSONPlaceholder é um serviço online gratuito que fornece APIs REST falsas para testes.
url_api = "https://jsonplaceholder.typicode.com/todos/1"

print(f"\nFazendo requisição GET para API: {url_api}")

try:
    # Faz a requisição GET para a URL da API
    # --- Camada Simples de Cibersegurança para Requisições Web ---
    # Novamente, timeout e verificação SSL são importantes.
    # Ao trabalhar com APIs, é importante validar o formato dos dados recebidos (JSON neste caso).
    response_api = requests.get(url_api, timeout=10) # Timeout de 10 segundos

    # Verifica se a requisição foi bem-sucedida
    if response_api.status_code == 200:
        print("Requisição à API bem-sucedida (Status Code 200)")
        try:
            # response.json() converte a resposta JSON em um dicionário Python
            # --- Camada Simples de Cibersegurança para Requisições Web ---
            # Tentar decodificar o JSON e capturar exceções de JSONDecodeError
            # ajuda a lidar com respostas malformadas ou inesperadas.
            dados_json = response_api.json()
            print("Dados JSON recebidos:")
            # Exibe os dados recebidos em um formato legível
            import json
            print(json.dumps(dados_json, indent=4)) # Usa json.dumps para formatar a saída
        except json.JSONDecodeError:
            print("Erro: Resposta da API não é um JSON válido.")
            print("Conteúdo da resposta (primeiros 200 caracteres):")
            print(response_api.text[:200] + "...")


    else:
        print(f"Erro na requisição à API. Status Code: {response_api.status_code}")

except requests.exceptions.Timeout:
    print(f"Erro na requisição à API: Tempo limite excedido para {url_api}")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro durante a requisição à API: {e}")
````
